- هدف از پرسیدن: بررسی پیامدهای حل‌نشدن مسئله.

پاسخ:

### سؤال 8: یک Business Objective Statement رسمی برای این پروژه بنویسید.

**هدف از پرسیدن:** ارائه یک تعریف رسمی و خلاصه از هدف اصلی کسب‌وکار.

**پاسخ:**

هدف این پروژه، شناسایی مشتریان در معرض ریزش در یک فروشگاه اینترنتی با استفاده از روش‌های داده‌کاوی است. ابتدا با استفاده از الگوریتم‌های خوشه‌بندی، الگوهای رفتاری مشتریان بررسی شده و گروه هدف شناسایی می‌شود. سپس با استفاده از الگوریتم‌های طبقه‌بندی، احتمال تعلق مشتریان به گروه هدف پیش‌بینی شده و در نهایت ۵۰۰ مشتری با بالاترین احتمال ریزش انتخاب و رتبه‌بندی می‌شوند. نتایج پروژه به فروشگاه کمک می‌کند تا برنامه‌های بازاریابی هدفمندتری اجرا کرده و برای حفظ مشتریان اقدام کند.

### سؤال 19: آیا این پروژه Classification، Regression، Clustering یا Association Rule است؟ دلایل خود را توضیح دهید.

**هدف از پرسیدن:** مشخص کردن نوع مسئله داده‌کاوی و دلیل انتخاب الگوریتم‌ها.

**پاسخ:**

این پروژه ترکیبی از دو نوع مسئله داده‌کاوی است:

1. **Clustering:** چون در دیتاست برچسب مشخصی برای مشتریان در معرض ریزش وجود ندارد، ابتدا مشتریان براساس ویژگی‌های رفتاری گروه‌بندی می‌شوند تا الگوهای مشابه شناسایی شوند.
2. **Classification:** پس از تحلیل خوشه‌ها و ساخت برچسب اولیه، از الگوریتم طبقه‌بندی برای پیش‌بینی احتمال تعلق مشتریان به گروه هدف استفاده می‌شود.

در نهایت مشتریان براساس احتمال پیش‌بینی‌شده رتبه‌بندی می‌شوند و ۵۰۰ نفر اول انتخاب خواهند شد.

| Classification | F1-Score | بررسی تعادل Precision و Recall |

### سؤال 21: Accuracy چیست و آیا برای این پروژه کافی است؟

**هدف از پرسیدن:** بررسی محدودیت استفاده از دقت کلی برای ارزیابی مدل.

**پاسخ:**

### سؤال 22: در این پروژه Precision مهم‌تر است یا Recall؟ چرا؟

**هدف از پرسیدن:** تعیین معیار مهم‌تر با توجه به هدف کسب‌وکار.

**پاسخ:**

در این پروژه هر دو معیار اهمیت دارند. Recall بالا باعث می‌شود تعداد بیشتری از مشتریان در معرض ریزش شناسایی شوند. Precision بالا نیز کمک می‌کند مشتریانی که به‌اشتباه در گروه هدف قرار گرفته‌اند کمتر باشند. با توجه به محدودیت انتخاب فقط ۵۰۰ مشتری، Precision و کیفیت رتبه‌بندی نیز اهمیت زیادی دارند.

### سؤال 23: موفقیت مدل را چگونه تعریف می‌کنید؟

**هدف از پرسیدن:** تعیین شرایطی که نشان می‌دهد مدل عملکرد قابل قبولی دارد.

**پاسخ:**

موفقیت مدل زمانی است که در مرحله اول بتواند گروه‌بندی مناسبی از مشتریان ایجاد کند و در مرحله دوم مشتریان گروه هدف را با عملکرد قابل قبول شناسایی و براساس احتمال پیش‌بینی‌شده به‌درستی رتبه‌بندی کند.

### سؤال 24: یک Data Mining Goal Statement رسمی بنویسید.

**هدف از پرسیدن:** ارائه تعریف رسمی از هدف فنی داده‌کاوی.

**پاسخ:**

در این پروژه ابتدا با استفاده از الگوریتم‌های خوشه‌بندی، مشتریان براساس ویژگی‌های رفتاری و الگوهای خرید گروه‌بندی می‌شوند. سپس با تحلیل ویژگی‌های هر خوشه، گروه مشتریان در معرض ریزش شناسایی و برچسب اولیه برای آن‌ها ایجاد می‌شود. در مرحله بعد، یک مدل طبقه‌بندی با استفاده از این برچسب‌ها آموزش داده می‌شود تا احتمال تعلق مشتریان به گروه هدف را پیش‌بینی کند. در نهایت، ۵۰۰ مشتری با بالاترین احتمال انتخاب و رتبه‌بندی می‌شوند.

### سؤال 25: رابطه بین هدف داده‌کاوی و هدف کسب‌وکار را توضیح دهید.

**هدف از پرسیدن:** نشان دادن ارتباط بین روش فنی پروژه و نتیجه مورد انتظار فروشگاه.

**پاسخ:**

هدف کسب‌وکار، کاهش ریزش مشتریان و حفظ مشتریان فعلی است. برای رسیدن به این هدف، ابتدا با Clustering الگوهای رفتاری مشتریان شناسایی می‌شوند و سپس با Classification مشتریان در معرض ریزش پیش‌بینی و رتبه‌بندی می‌شوند. خروجی این فرایند به فروشگاه کمک می‌کند تا اقدامات بازاریابی خود را روی ۵۰۰ مشتری منتخب متمرکز کند.
